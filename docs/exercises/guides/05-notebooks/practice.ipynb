## 1. Notebook state follows execution order

A kernel is a Python process with variables in memory. The order you execute cells determines the values. Restarting removes that state.

**Predict:** Will changing fruit rebuild the previously formed string?

```python
fruit = "bananas"
sentence = f"I like {fruit}."
fruit = "apples"
print("Old string:", sentence)
print("New string:", f"I like {fruit}.")
assert sentence == "I like bananas."
```

**Check:** No. The old string holds its earlier value. Re-execute its construction to rebuild it.

## 2. Working directory vs. source directory

Relative paths start at the process working directory. They do not automatically start next to a notebook. This example inspects paths only and never creates or deletes files.

**Predict:** What folder is the starting point for Path(".")?

```python
from pathlib import Path
import sys
current = Path.cwd()
relative = Path("docs")
print("Working directory:", current)
print("Resolved relative path:", relative.resolve())
print("Interpreter:", sys.executable)
assert relative.resolve().is_absolute()
```

**Check:** The kernel/process working directory. An absolute path identifies the intended location independently of cd.

**Change one thing:** Compare the path from the repository root with the path from a week folder. Explain why cd changes relative-file behavior.

## 3. Streaming sends partial progress

A generator yields multiple updates. A streamed API can include multiple event types, so a UI should select the text events rather than assume every event is text.

**Predict:** How many text chunks are selected?

```python
events = [
    {"type": "status", "value": "started"},
    {"type": "text", "value": "Hello "},
    {"type": "text", "value": "learner"},
]
def text_chunks(items):
    for event in items:
        if event["type"] == "text":
            yield event["value"]
chunks = list(text_chunks(events))
print(chunks, "→", "".join(chunks))
assert "".join(chunks) == "Hello learner"
```

**Check:** Two. yield preserves progress; return ends an ordinary function.

**Change one thing:** Insert a non-text event between the chunks. Confirm it does not become displayed answer text.
