## 1. Evaluate a small prototype with a fixed dataset

Choose a success measure before comparing approaches. Use a small curated set so you can reproduce failures. Scores below describe fictional fixtures, not real model performance.

**Predict:** Which fixture scores better, and what does the score actually measure?

```python
expected = ["4", "Paris", "unknown"]
candidates = {"prototype_a": ["4", "Paris", "invented"],
              "prototype_b": ["4", "Paris", "unknown"]}
scores = {name: sum(answer == target for answer, target in zip(values, expected)) / len(expected)
          for name, values in candidates.items()}
print(scores)
assert scores["prototype_b"] == 1
```

**Check:** prototype_b on these fixtures. A small exact-match dataset does not establish general quality.

**Change one thing:** Add a case where a correct paraphrase differs from the expected string. Explain why exact-match scoring may be insufficient.

## 2. Reproduce, inspect, simplify, fix

A small failing case separates causes. Catch the exact exception in a teaching example so the notebook can continue, then implement a specific repair.

**Predict:** Will the original operation raise NameError or KeyError?

```python
record = {"title": "Agents"}
try:
    print(record["author"])
except KeyError as error:
    print("Reproduced:", type(error).__name__)
author = record.get("author", "Unknown")
print("Fixed optional lookup:", author)
assert author == "Unknown"
```

**Check:** KeyError because the dictionary exists but the key does not. NameError means the variable itself was not defined.

**Change one thing:** Remove the record definition in a scratch cell. Explain how the new NameError differs from the original missing-key error.

## 3. Generated code is an output before it is execution

A coding workflow designs, writes, and reviews code. Generated files or Docker execution are side effects. First inspect and validate a harmless code string locally.

**Predict:** Does parsing Python text execute it?

```python
import ast
proposed = "def add(a, b):\n    return a + b\n"
tree = ast.parse(proposed)  # Parse syntax only; do not execute generated text.
functions = [node.name for node in tree.body if isinstance(node, ast.FunctionDef)]
print(functions)
assert functions == ["add"]
checks = {"has_function": True, "has_test_plan": False}
print("Review:", checks)
```

**Check:** No. Syntax validity does not prove behavior or make generated code safe to execute.

**Change one thing:** Write a test plan for the intended function. Explain which separate step would run code and where the source uses Docker.
