## 1. Inspect a snapshot before recording changes

Git distinguishes working files, staged changes, and committed snapshots. This model teaches the distinction without modifying repository history.

**Predict:** Which changes are recorded when only notes.md is staged?

```python
working = {"notes.md": "new explanation", "app.py": "unreviewed edit"}
staged = {"notes.md": working["notes.md"]}
snapshot = dict(staged)
print("Working:", list(working), "Staged:", list(staged), "Commit:", list(snapshot))
assert "app.py" not in snapshot
```

**Check:** Only notes.md. git diff --cached shows staged content; pushing sends existing commits.

**Change one thing:** Use git status and git diff in PowerShell to inspect real state. Stage only files you intentionally want in a commit.

## 2. Working directory vs. source directory

Relative paths start at the process working directory. They do not automatically start next to a notebook. This example inspects paths only and never creates or deletes files.

**Predict:** What folder is the starting point for Path(".")?

```python
from pathlib import Path
import sys
current = Path.cwd()
relative = Path("docs")
print("Working directory:", current)
print("Resolved relative path:", relative.resolve())
print("Interpreter:", sys.executable)
assert relative.resolve().is_absolute()
```

**Check:** The kernel/process working directory. An absolute path identifies the intended location independently of cd.
